🌟 Exercise 1: Rewrite and Optimize a Vague Prompt
Scenario:
You’re an AI Prompt Engineer at a productivity startup. Your manager asks you to use ChatGPT to create a LinkedIn post promoting their new focus app, FlowNest. They send you this vague prompt:

“Write something about productivity tips.”



Tasks
Identify 3 critical issues with this prompt.
(Hint: Think about brand voice, target users, and platform engagement.)

Rewrite the prompt using all three techniques:

Role prompting (e.g., “Act as a…”)
Instruction-based prompting (e.g., “Give 3 actionable…”)
Format/tone/length constraint (e.g., “Use bullet points” or “max 280 characters”)

Add a final line that makes the prompt reusable (e.g., “Use this format whenever we promote productivity tools on social media”).

```python
# Step 1: Identify 3 critical issues with the original prompt (“Write something about productivity tips.”)
    # No clear brand voice                  → The prompt doesn’t reflect FlowNest’s identity.
    # No clear target audience              → It doesn’t specify if the post is for students, professionals, or remote workers.
    # No clear platform engagement guidance → It ignores LinkedIn’s style (professional, concise, engagement-driven).
```

```python
# Step 2 : Rewrite the prompt using all three techniques:
    # Role prompting (e.g., “Act as a…”)
    # Instruction-based prompting (e.g., “Give 3 actionable…”)
    # Format/tone/length constraint (e.g., “Use bullet points” or “max 280 characters”)

    # Improved Prompt:
      # Act as a productivity coach writing a LinkedIn post for FlowNest, our new focus app.
      # Give 3 actionable productivity tips tailored to busy young professionals.
      # Use a friendly, professional tone, format with bullet points, and keep it under 200 words to match LinkedIn style.
      # End the post with a short call-to-action encouraging readers to try FlowNest.
      # Use this format whenever we promote productivity tools on social media --> cf. step3 : Add a final line that makes the prompt reusable (e.g., “Use this format whenever we promote productivity tools on social media” )
```

```python
# Exercise 2: Multi-Part Prompt for Quiz Generation
    # Scenario: a 7th-grade science teacher uploads a short article on volcanic eruptions. You are tasked with writing a prompt for ChatGPT to create engaging educational content.

    # Tasks
      # 1. Write a complete prompt that asks the model to:
        # Summarize the article in 2 simple bullet points
        # Generate 3 age-appropriate multiple-choice questions, each with 1 correct and 2 distractors
        # Include a friendly tone and simple vocabulary for 11–13-year-olds
        # Output everything in a format suitable for Google Slides

      # Proposed prompt :
        # "Act as an educational content creator for middle school students. You are given a short article about volcanic eruptions.
        # Please create an engaging quiz presentation in a format ready for Google Slides.
        # Summarize the article in 2 simple bullet points using friendly language suitable for 11–13-year-olds.
        # Generate 3 multiple-choice questions based on the article. Each question should:
          # Be clear and age-appropriate
          # Have 1 correct answer and 2 distractors
          # Use simple vocabulary
        # Organize the output in Google Slides format:
          # Slide 1: Title (“Volcanic Eruptions – Key Ideas”) + the 2 summary bullet points
          # Slides 2–4: Each slide shows 1 question, the 3 answer choices, and highlight the correct answer clearly.
          # Keep the tone friendly, supportive, and easy to understand. Avoid overly technical terms. Use formatting that a teacher could copy directly into Google Slides."*

      # 2. Compare your prompt to this vague one: “Make a quiz for kids about this article.”
        # Explain three key ways your version prevents irrelevant or low-quality output :
          # Clarity of tasks          → Instead of “make a quiz,” I broke down tasks (summarize + create 3 MCQs + format for slides). This ensures ChatGPT covers everything.
          # Defined audience and tone → I specified 11–13-year-olds, simple vocabulary, friendly tone, which prevents the model from outputting text that is too complex or too formal.
          # Structured output format  → I asked for Google Slides-ready sections (slide titles, bullet points, clear separation). This prevents messy text and ensures the output is usable right away.
```

```python
# Step 2: Rewrite the intern’s prompt
    # Improved prompt : includes at least 4 context types (role, audience, purpose, format, constraints):
        # "Act as a financial analyst preparing a monthly finance update. Summarize the attached financial report for a non-technical executive team. The purpose is a 3-minute spoken presentation. Please provide exactly 3 key takeaways, each supported by one or two important financial figures (e.g., revenue growth, cost changes, forecasts). Format the output as Google Slides-style bullet points:
        # Slide 1: Title + overview sentence
        # Slides 2–4: One key takeaway per slide, with supporting data
        # Keep the language clear, concise, and free of jargon. Limit total length to about 200–250 words."*

# Comments : Why this works
        # It prevents generic output because the AI now knows its role (financial analyst).
        # It ensures relevance because the audience and purpose are defined.
        # It forces useful numbers by requiring supporting data.
        # It guarantees usable output by specifying slides format + word limit.
```

```python
# Exercise 4: Match Prompt to Purpose
    # Scenario: You’re building prompt templates for a customer support chatbot at an e-commerce company. Pick one of the following styles and use it to write a complete prompt for the chatbot.
        # Exploratory: Helps the model brainstorm options or generate ideas
        # Structured: Follows a fixed format (lists, sections, steps)
        # Conversational: Mimics a human, chatty tone
        # Functional: Executes a task with precision (e.g., summarizing or transforming input)

    # Step 1 - write your prompt using the selected style.
        # I’ll choose the Conversational style, since it fits very well with a customer support chatbot that needs to sound friendly and approachable.
        # Proposed prompt (Conversational Style):
            # "You are a helpful and friendly customer support chatbot for an e-commerce website. When customers ask questions about their orders, products, or returns, respond in a warm, conversational tone.
            # Greet the customer by acknowledging their concern.
            # Answer their question clearly and politely, avoiding technical jargon.
            # If the solution requires steps, explain them one by one in simple language.
            # Offer additional help at the end (e.g., 'Would you like me to track your package for you?').
            # Keep answers short (3–4 sentences), supportive, and written as if you are chatting with a real person."
    # Step 2 - identify 2 key features that make your prompt a match for the style.
        # Natural and friendly tone   → The chatbot is told to sound warm, friendly, and human-like instead of robotic.
        # Engagement and empathy      → The prompt instructs the model to greet, acknowledge, and offer extra help, which makes the conversation flow like real customer service.
    # Step 3 - Justify your choice; why is this style better than the others for your specific case?
        # The Conversational style is the best fit for a customer support chatbot because:
            # Customers expect a human-like, empathetic interaction when they have problems, not a rigid list or technical summary.
            # It builds trust and satisfaction, which is critical in e-commerce, where issues often involve money, refunds, or delayed deliveries.
            # Compared to Structured or Functional, it feels more personal; compared to Exploratory, it stays focused and efficient.
```

```python
# Exercise 6: Hallucination Spotting and Mitigation
    # Scenario: you’re preparing a summary of a peer-reviewed article on climate change and marine biodiversity. ChatGPT inserts this claim:
        # “Over 50% of marine species are projected to go extinct by 2050.”
        # This claim does not appear in the original article.

    # Step 1. Write a precise prompt instructing the model to avoid hallucinations and use only the provided text.
        # Proposed prompt (sticking only to the text) :
            # "Summarize the following peer-reviewed article on climate change and marine biodiversity. Use only the information explicitly present in the text.
            # Do not add outside knowledge, assumptions, or statistics that are not stated in the article."
    # Step 2. Rewrite the prompt using verification language, like:
        # “Only include claims explicitly stated in the article" / “If uncertain, state that the information is not available.”
        # Proposed alternative prompt (including verification instructions to actively prevent hallucination) :
            # "Summarize the following peer-reviewed article on climate change and marine biodiversity. Only include claims and figures explicitly stated in the article.
            # If uncertain or if information is missing, clearly state: ‘This information is not available in the article.’
            # Do not invent or infer numbers, projections, or conclusions beyond the text."
    # Step 3. List 2 mitigation strategies you used to reduce hallucination risk.
            # Instructional constraints → Explicitly told the model to only use claims from the source text.
            # Uncertainty handling      → Gave the model a fallback phrase (“This information is not available in the article”) so it won’t fabricate missing details.
    # Step 4. Name 2 professional domains (e.g., legal, healthcare) where hallucinations are especially risky, and explain why.
            # Legal domain → Hallucinated legal precedents, case law, or statutes could mislead lawyers, affect contracts, or even invalidate court arguments. Accuracy and citation fidelity are critical.
            # Healthcare/medical domain → Fabricated medical data, diagnoses, or treatment recommendations could directly endanger patient health, leading to harmful or even life-threatening consequences.



```
