```python
# Scenario
    # Imagine you’re building a personalized welcome tool for a community platform. You want users to feel greeted warmly whenever they log in. Your app will:
        # Take their name as input
        # Detect the current time
        # Return a personalized greeting message
        # Bonus: Let users choose the greeting language (English or French).
```

```python
from datetime import datetime

def time_based_greeting(name: str, language: str = "en", now: datetime | None = None) -> str:
    """
    Return a time-aware greeting in English or French.

    Args:
        name: The person's name (e.g., "Alice").
        language: "en" for English or "fr" for French. Defaults to "en".
        now: Optional datetime for testing; if None, uses current local time.

    Behavior:
        - English:
            05:00–11:59  -> "Good morning, {name}!"
            12:00–17:59  -> "Good afternoon, {name}!"
            18:00–04:59  -> "Good evening, {name}!"
        - French:
            05:00–17:59  -> "Bonjour, {name}!"
            18:00–04:59  -> "Bonsoir, {name}!"
        - Unknown language -> falls back to English.
    """
    now = now or datetime.now()
    hour = now.hour

    # Determine time-of-day buckets
    if 5 <= hour < 12:
        tod = "morning"
    elif 12 <= hour < 18:
        tod = "afternoon"
    else:
        tod = "evening"  # includes evening + night

    # Language-specific phrasing
    if language.lower() == "fr":
        # In French, "Bonjour" until ~18:00, then "Bonsoir"
        greeting = "Bonjour" if tod in ("morning", "afternoon") else "Bonsoir"
        return f"{greeting}, {name}!"
    else:
        # Default to English
        if tod == "morning":
            return f"Good morning, {name}!"
        elif tod == "afternoon":
            return f"Good afternoon, {name}!"
        else:
            return f"Good evening, {name}!"
```

```python
# How to Use
print(time_based_greeting("Alice", "en"))  # e.g., "Good morning, Alice!"
print(time_based_greeting("Alice", "fr"))  # e.g., "Bonjour, Alice!" or "Bonsoir, Alice!"
```
