```python
# Exercise 1: Exercise 1 : Gradio Interface - Multi-Function Toolkit

import gradio as gr

# Example functions
def greet(name):
    return f"Hello, {name}!"

def square(x):
    return x ** 2

# Build interfaces
greet_interface = gr.Interface(fn=greet, inputs=gr.Textbox(label="Name"), outputs="text")
square_interface = gr.Interface(fn=square, inputs=gr.Number(label="Enter a number"), outputs="number")

# Combine into a tabbed interface
demo = gr.TabbedInterface(
    [greet_interface, square_interface],
    ["Greet", "Square"]
)

demo.launch()
```

```python
# Exercise 2: Gradio Blocks – Simple Two-Step Calculator
# Requirements covered:
# - gr.Blocks, gr.Row layout
# - Inputs: gr.Number, gr.Radio
# - Action: gr.Button
# - Output: gr.Label
# - Markdown instructions
# - Event wiring via .click()
# - Conditional logic in backend function

import gradio as gr

def calculate(a, b, operation):
    # Basic validation (Gradio Number returns None if empty)
    if a is None or b is None:
        return "Please enter both numbers."

    try:
        a = float(a)
        b = float(b)
    except ValueError:
        return "Inputs must be numeric."

    if operation == "Add":
        result = a + b
    elif operation == "Multiply":
        result = a * b
    else:
        return "Choose an operation."

    # gr.Label accepts a string or number. We return a friendly string.
    return f"Result: {result:g}"

with gr.Blocks(title="Two-Step Calculator") as demo:
    gr.Markdown(
        """
# Two-Step Calculator (Gradio Blocks)
Enter two numbers, select **Add** or **Multiply**, then click **Calculate**.
- Uses **gr.Blocks** with **gr.Row()** for custom layout
- Inputs: `Number`, `Radio`
- Action: `Button`
- Output: `Label`
        """
    )

    with gr.Row():
        num1 = gr.Number(label="First Number", value=None, interactive=True)
        num2 = gr.Number(label="Second Number", value=None, interactive=True)

    op = gr.Radio(
        choices=["Add", "Multiply"],
        value="Add",
        label="Operation",
        interactive=True
    )

    calc_btn = gr.Button("Calculate")
    output = gr.Label(label="Output")

    calc_btn.click(
        fn=calculate,
        inputs=[num1, num2, op],
        outputs=output
    )

if __name__ == "__main__":
    demo.launch()
```

```python
# Exercise 3 : Streamlit Chat - Stateful Memory Bot
# Requirements covered:
# - st.chat_message() chat bubbles
# - st.session_state["chat_history"] to persist conversation
# - Loop through history to render past messages
# - Append user + bot messages on new input
# - "Clear Chat" button to reset

%pip install streamlit
import streamlit as st

st.set_page_config(page_title="Stateful Memory Bot", page_icon="💬")

# ---- Initialize session state ----
if "chat_history" not in st.session_state:
    st.session_state.chat_history = [
        {"role": "assistant", "content": "Hi! I remember our conversation. How can I help today?"}
    ]

# ---- Header + Clear button ----
col1, col2 = st.columns([1, 0.2])
with col1:
    st.markdown("## 💬 Streamlit Chat — Stateful Memory Bot")
with col2:
    if st.button("Clear Chat", use_container_width=True):
        st.session_state.chat_history = [
            {"role": "assistant", "content": "Chat cleared. How can I help now?"}
        ]
        st.experimental_rerun()

st.caption("This demo uses `st.chat_message()` and `st.session_state` to keep memory across turns.")

# ---- Render previous messages ----
for msg in st.session_state.chat_history:
    with st.chat_message(msg["role"]):
        st.markdown(msg["content"])

# ---- Chat input (user) ----
user_input = st.chat_input("Type your message...")
if user_input:
    # Append user message
    st.session_state.chat_history.append({"role": "user", "content": user_input})
    with st.chat_message("user"):
        st.markdown(user_input)

    # ---- Simple echo bot reply (placeholder for real logic) ----
    bot_reply = f"You said: {user_input}"
    st.session_state.chat_history.append({"role": "assistant", "content": bot_reply})

    with st.chat_message("assistant"):
        st.markdown(bot_reply)
```

```python
# Exercise 4 : Streamlit UI + Logic Combo
# Requirements covered:
# - st.title, st.radio layout
# - st.line_chart with NumPy-generated data
# - Conditional rendering (Show Code / Show JSON)
# - st.code and st.json
# - Bonus: st.image from a public URL

import streamlit as st
import numpy as np
import pandas as pd
from datetime import datetime, timedelta

st.set_page_config(page_title="Mini Dashboard", page_icon="📊", layout="centered")

# ---- Title ----
st.title("📊 Mini Dashboard – Dynamic Content")

# ---- Random data (NumPy) for the line chart ----
np.random.seed(42)  # for reproducibility
days = 30
dates = [datetime.today() - timedelta(days=i) for i in range(days)][::-1]
data = np.random.randn(days, 3).cumsum(axis=0)  # 3 series, cumulative to look like trends
df = pd.DataFrame(data, index=pd.to_datetime(dates), columns=["Series A", "Series B", "Series C"])

st.subheader("Random Line Chart (Generated with NumPy)")
st.line_chart(df)

# ---- Radio selector ----
choice = st.radio("Choose what to display below:", ["Show Code", "Show JSON"], horizontal=True)

# ---- Conditional rendering ----
sample_code = '''\
# Example: simple function and loop
def greet(name: str) -> str:
    return f"Hello, {name}!"

for who in ["Alice", "Bob", "Charlie"]:
    print(greet(who))
'''

sample_json = {
    "product": "PulseOne Mini",
    "version": "1.0.3",
    "metrics": {
        "active_users": 1275,
        "error_rate": 0.004,
        "avg_session_minutes": 7.8
    },
    "features": ["fitness-tracking", "bluetooth", "long-battery"],
    "timestamp": datetime.utcnow().isoformat() + "Z"
}

if choice == "Show Code":
    st.subheader("🔧 Code Snippet")
    st.code(sample_code, language="python")
else:
    st.subheader("🧾 JSON Preview")
    st.json(sample_json)

# ---- Bonus: Public image ----
st.markdown("---")
st.subheader("Bonus: Public Image")
st.image(
    "https://images.unsplash.com/photo-1519389950473-47ba0277781c",
    caption="Dashboard vibes (Unsplash)",
    use_column_width=True,
)
```

```python
# Exercise 5 : FastAPI - Smart Responder
    # Build a FastAPI backend app that returns different responses based on user input

# Install dependencies (Notebook-only)
%pip install fastapi uvicorn pyngrok python-dotenv

# Imports and basic setup (FastAPI, models, server, tunneling, env)
# - FastAPI: web framework
# - BaseModel (Pydantic): parse/validate JSON bodies
# - uvicorn: ASGI server to run FastAPI
# - pyngrok: create a public URL to your local server
# - nest_asyncio: allow nested event loops (useful in notebooks)
# - dotenv/os: load NGROK token from .env or environment variables

from fastapi import FastAPI
from pydantic import BaseModel
import uvicorn
from pyngrok import ngrok
import nest_asyncio
import os
from dotenv import load_dotenv

# Enable nested event loops for notebooks (prevents "RuntimeError: This event loop is already running")
nest_asyncio.apply()

# Create the FastAPI application
app = FastAPI(title="Smart Responder")

# Define the request schema with Pydantic
# Expect JSON like: {"message": "What's 5+5?"}

class Query(BaseModel):
    message: str

# Define the /respond POST endpoint
# - Lowercase the text
# - Simple routing based on keywords
# - Return different messages as JSON

@app.post("/respond")
def respond(payload: Query):
    text = (payload.message or "").lower()

    if "math" in text:
        return {"response": "Using calculator tool..."}
    if "date" in text:
        return {"response": "Fetching current date..."}
    return {"response": "Default LLM response."}
```
