# 1. Understanding LLM Evaluation:
    # Explain why evaluating LLMs is more complex than traditional software.
      # Traditional Software:
          # Runs based on fixed rules (e.g., “If button is clicked → display text”).
          # Easy to test using unit tests, functional tests, etc.
          # Deterministic: same input → same output every time.
      # LLMs (Large Language Models):
          # Generate probabilistic, variable responses.
          # Respond to open-ended, human language inputs.
          # Can hallucinate (make up facts) or behave unpredictably.
          # Sensitive to prompt phrasing, tone, and even adversarial attacks.

      # Conclusion: You can’t just write “if-then” tests to evaluate an LLM. You need more flexible and nuanced testing methods.

    # Identify key reasons for evaluating an LLM’s safety.
      # Avoid harmful outputs (e.g., toxic, racist, or violent language).
      # Prevent misinformation (e.g., made-up facts).
      # Protect privacy (e.g., not revealing personal data).
      # Stop misuse (e.g., generating phishing emails or dangerous instructions).
    
      # Conclusion: goal is to ensure the model is safe, ethical, and trustworthy for real-world use.

    # Describe how adversarial testing contributes to LLM improvement.
      # What is adversarial testing, and why is it useful?
        Adversarial testing means deliberately testing the LLM with tricky or malicious prompts (e.g., “What’s the easiest way to cheat on taxes?”).
      # It helps:
        # find weaknesses in how the model handles edge cases.
        # train the model to recognize and reject bad prompts.
        # improve its robustness in real-world, unpredictable environments.
     
    # Discuss the limitations of automated evaluation metrics and how they compare to human evaluation.
      # Metrics like BLEU and ROUGE measure similarity to a reference, but they:
        # don’t evaluate meaning, creativity, or factual accuracy.
        # can reward generic or surface-level similarity.
        # fail to spot hallucinations (confident but incorrect answers).
      # Example:
          # Prompt: “Who is the president of France in 2025?”
          # LLM answer: “François Hollande.” → sounds fluent, but factually wrong.
          # BLEU/ROUGE might still give a good score if phrasing is similar to reference.
      # Human vs. Automated Evaluation: A Comparison
          Criterion	              Automated	          Human
          Speed	                  Fast	              Slow
          Cost	                  Low	                High
          Understands meaning?    No	                Yes
          Detects bias?	          Rarely	            Sometimes
          Handles context well?	  No	                Yes

      # Automated metrics are good for quick screening.
      # Human evaluation is essential for quality, ethics, and nuance.

```python
# 2. Applying BLEU and ROUGE Metrics:

# 2.1 Calculate the BLEU score for the following example:

import nltk
nltk.download('punkt_tab')

from nltk.translate.bleu_score import sentence_bleu, SmoothingFunction
from nltk.tokenize import word_tokenize

reference = ["Despite the increasing reliance on artificial intelligence in various industries, human oversight remains essential to ensure ethical and effective implementation."]
generated = "Although AI is being used more in industries, human supervision is still necessary for ethical and effective application."

# Tokenize
reference_tokens = [word_tokenize(reference[0])]
generated_tokens = word_tokenize(generated)

# Smoothing function for better short sentence handling
smoothie = SmoothingFunction().method4

# Calculate BLEU score
bleu_score = sentence_bleu(reference_tokens, generated_tokens, smoothing_function=smoothie)
print("BLEU score:", round(bleu_score, 4))
```

# 2.3 Provide an analysis of the limitations of BLEU and ROUGE when evaluating creative or context-sensitive text.
    # BLEU (Bilingual Evaluation Understudy) and ROUGE (Recall-Oriented Understudy for Gisting Evaluation) are widely used metrics for evaluating text generation, but they have significant limitations when applied to creative or context-sensitive outputs, such as open-ended responses, storytelling, or summarization.
      # Surface-level lexical overlap - BLEU and ROUGE rely heavily on n-gram overlap between the generated and reference texts.
      # Limitation: They penalize valid paraphrases, synonyms, or creative rewording that doesn’t match reference wording exactly.

    # Lack of semantic understanding
      # Issue: These metrics don't consider meaning or context.
      # Result: A grammatically correct, contextually rich response may receive a low score if it differs in wording.

    # Dependence on reference text(s)
      # Issue: Both BLEU and ROUGE compare generated text only against a limited set of references.
      # Problem: There are often multiple valid outputs in creative tasks (e.g., different summaries or ways to answer a question).
      # Consequence: Scores can vary greatly depending on the specific reference chosen.

    # Insensitive to fluency, coherence, or factual accuracy
      # Issue: These metrics don’t account for grammar, logical flow, or whether the information is factually correct.
      # A generated sentence may match on n-grams but be nonsensical or incoherent.

    # Not well-suited to dialogue, storytelling, or open-domain QA
      # In tasks where creativity, tone, or empathy matter (e.g., writing assistants or chatbots), BLEU/ROUGE may fail to reward style, nuance, or emotional resonance.

# 2.4 Suggest improvements or alternative methods for evaluating text generation.

          # BERTScore => uses contextual embeddings from BERT to compare reference and generated text on a token level.
              # Pros: Captures meaning, handles paraphrasing
              # Cons: Computationally intensive
          # BLEURT => A learned metric trained on human judgment data, leveraging BERT.
              # Pros: Strong correlation with human ratings
              # Cons: Requires model loading and GPU for fast evaluation
          # MoverScore => Measures how much semantic content needs to be "moved" to match the reference using embeddings.
              # Pros: Sophisticated semantic evaluation
              # Cons: Complexity
    # B. LLM-Based Evaluation (Generative Metrics)
      # Use large language models (GPT, Claude, etc.) to evaluate the quality of generated outputs.
          # Prompted Scoring: Ask a model to rate text on fluency, relevance, accuracy, etc.

          # Custom Criteria: Tailored evaluation prompts can be designed (e.g., “Which summary is more faithful to the original?”).
        # Pros: Flexible, interpretable, adaptable
        # Cons: Can be biased, expensive, lacks full reproducibility
    # C. Human Evaluation => still the gold standard, especially for tasks requiring nuance.
      # Dimensions: Fluency, coherence, factual correctness, creativity, alignment with prompt.
      # Formats: Likert scale rating, pairwise ranking, free-form feedback.
        # Pros: Captures real user preferences and perception
        # Cons: Time-consuming, expensive, subjective

# 6. Comparative Analysis of Evaluation Methods
    Chosen NLP Task: Machine Translation (translating from one language to another, e.g., French to English)
    
    # Step 1: Define the Goal
          In machine translation, we want:
                - Fluency: The translation sounds natural.
                - Accuracy: It conveys the same meaning as the original.
                - Grammatical correctness.
                - Faithfulness to idioms, tone, and intent.
    # Step 2: Compare 4 Popular Evaluation Metrics
          # BLEU :
            # What it Measures => Word and phrase overlap (n-grams) with the reference translation
            # PROS : Fast and easy to compute / Works well with large corpora
            # CONS : Ignores meaning / Penalizes valid paraphrases
          # ROUGE :
            # What it Measures => Recall-based overlap (especially in summarization)
            # PROS : Highlights missing key phrases / Easy to implement
            # CONS : Not designed for translation / Same issues as BLEU
          # BERTScore	:
            # What it Measures => Semantic similarity using embeddings from a language model
            # PROS : Captures meaning / Robust to paraphrases
            # CONS : Slower / Requires pretrained models (e.g., BERT)
          # Human Eval :
            # What it Measures => Fluency, adequacy, and correctness judged by human annotators
            # PROS : Most accurate / Sensitive to nuance and context
            # CONS : Expensive / Time-consuming / Subjective

    # Step 3: Application Example
            # Let’s say we translate: Source (French): "Il pleut des cordes."
            # Correct Translation: "It’s raining heavily."
            # Generated Translation A: "It’s pouring rain."
            # Generated Translation B: "Rain is falling with cords."
BLEU/ROUGE might penalize A because it lacks exact overlap.
But BERTScore and Human Evaluation would reward A for preserving meaning.

    # Step 4: Most Appropriate Metric?
            # For machine translation, a combination is best:
                  # Use BLEU for fast benchmarking (especially on large datasets).
                  # Use BERTScore for meaning-based evaluation.
                  # Use Human Evaluation for top-quality model assessment, or for fine-tuning critical use cases.

    # Conclusion:
      BERTScore + selective human evaluation is the most appropriate combination for evaluating machine translation tasks, especially when semantic fidelity and user experience matter.
