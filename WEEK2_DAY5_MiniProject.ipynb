```python
# 1. Setting data model and constants
from typing import List, Optional
Board = List[str]

WIN_COMBOS = [
    (0, 1, 2), (3, 4, 5), (6, 7, 8),  # rows
    (0, 3, 6), (1, 4, 7), (2, 5, 8),  # columns
    (0, 4, 8), (2, 4, 6)              # diagonals
]
```

```python
# 2. Create a ready-to-run Tic Tac Toe Python script and save it for download
code = r'''#!/usr/bin/env python3
'''
# -*- coding: utf-8 -*-
```

```python
# 3. Display the Tic Tac Toe board with numbers for empty cells.
def display_board(board: Board) -> None:
    cells = [c if c != " " else str(i+1) for i, c in enumerate(board)]
    print()
    print(f" {cells[0]} │ {cells[1]} │ {cells[2]} ")
    print(f"───┼───┼───")
    print(f" {cells[3]} │ {cells[4]} │ {cells[5]} ")
    print(f"───┼───┼───")
    print(f" {cells[6]} │ {cells[7]} │ {cells[8]} ")
    print()
```

```python
# 4. Prompt the current player for a move (1–9). Validate both format and rules. Return a 0-based index for the chosen empty cell. You can end the game here by typing: 'q', 'quit', or 'exit'.
def player_input(player: str, board: Board) -> int:

    while True:
        try:
            raw = input(
                f"Player {player}, choose a position (1-9) "
                f"or type 'q' to quit: "
            ).strip()
        except (EOFError, KeyboardInterrupt):
            print("\nQuitting game.")
            raise SystemExit(0)

        if raw.lower() in {"q", "quit", "exit"}:
            print("Quitting game.")
            raise SystemExit(0)

        if not raw.isdigit():
            print("Please enter a number between 1 and 9.")
            continue

        pos = int(raw)
        if not 1 <= pos <= 9:
            print("Out of range. Choose a number between 1 and 9.")
            continue

        idx = pos - 1
        if board[idx] != " ":
            print("That square is taken. Choose another.")
            continue

        return idx
```

```python
# 5. Return 'X' or 'O' if a player has won; otherwise None.

def check_win(board: Board) -> Optional[str]:
    """Return 'X' or 'O' if a player has won; otherwise None."""
    for a, b, c in WIN_COMBOS:
        if board[a] != " " and board[a] == board[b] == board[c]:
            return board[a]
    return None
```

```python
# 6. Return True if there are no empty spaces left on the board
def is_full(board: Board) -> bool:
    return all(cell != " " for cell in board)
```

```python
# 7. Return the next player's symbol
def switch(player: str) -> str:
    return "O" if player == "X" else "X"
```

```python
# 8. Ask if the players want another round. Return True to continue, False to terminate the program.
def ask_play_again() -> bool:
    try:
        again = input("Play again? (y/n): ").strip().lower()
    except (EOFError, KeyboardInterrupt):
        print("\nGoodbye.")
        return False

    return again in {"y", "yes", "oui"}
```

```python
# 9. Main loop coordinating setup, turns, win/tie detection, and replay :
def play() -> None:
    print("=== Tic Tac Toe ===")
    print("Two players (X and O). Enter 1-9 to place your mark.")
    print("End the game anytime with: 'q', 'quit', 'exit' or Ctrl+C.\n")

    while True:                      # session: one or more rounds
        board: Board = [" "] * 9     # round setup
        current = "X"
        display_board(board)

        while True:                  # one round
            idx = player_input(current, board)  # 1) get a valid move
            board[idx] = current                # 2) apply the move
            display_board(board)                # 3) show result

            winner = check_win(board)           # 4) check win
            if winner:
                print(f"🎉 Player {winner} wins!\n")
                break

            if is_full(board):                  # 5) check tie
                print("🤝 It's a tie!\n")
                break

            current = switch(current)           # 6) next player

        if not ask_play_again():                # 7) replay decision
            print("Thanks for playing! Goodbye.")
            return
```

```python
# 10. Program entry point and clean exit
if __name__ == "__main__":
    try:
        play()
    except SystemExit:
        print("Game terminated by user.")
```
