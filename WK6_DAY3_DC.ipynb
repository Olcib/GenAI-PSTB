```python
# Classify Handwritten Digits with CNNs

# Load the MNIST dataset:

# Installation de TensorFlow (si nécessaire) pour pouvoir charger le jeu de données MNIST
!pip install -q tensorflow

# Importation après installation
import numpy as np
import matplotlib.pyplot as plt
from tensorflow.keras.datasets import mnist
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Flatten, Conv2D, MaxPooling2D
from tensorflow.keras.utils import to_categorical
```

```python
# Étape 2 : Chargement du jeu de données MNIST

from tensorflow.keras.datasets import mnist

# Charger les données
(x_train, y_train), (x_test, y_test) = mnist.load_data()

# Afficher les dimensions
print("x_train shape :", x_train.shape)
print("y_train shape :", y_train.shape)
print("x_test shape  :", x_test.shape)
print("y_test shape  :", y_test.shape)
```

```python
# Étape 1 : Flatten
x_train_flat = x_train.reshape(x_train.shape[0], 28*28)
x_test_flat = x_test.reshape(x_test.shape[0], 28*28)

# Étape 2 : Normalisation
x_train_flat = x_train_flat / 255.0
x_test_flat = x_test_flat / 255.0

# Étape 3 : One-hot encoding
from tensorflow.keras.utils import to_categorical
y_train_cat = to_categorical(y_train, num_classes=10)
y_test_cat = to_categorical(y_test, num_classes=10)
```

```python
# Build and train a Fully Connected Neural Network:
    # Create a Sequential model.
    # Add Dense layers with appropriate activation functions (e.g., ReLU and softmax).
    # Compile the model with an optimizer (e.g., Adam), loss function (e.g., categorical cross-entropy), and metrics (e.g., accuracy).
    # Train the model using model.fit() and evaluate its performance.#

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.utils import to_categorical

# Créer et compiler le modèle
model_fcn = Sequential([
    Dense(128, activation='relu', input_shape=(784,)),
    Dense(64, activation='relu'),
    Dense(10, activation='softmax')
])

model_fcn.compile(optimizer='adam',
                  loss='categorical_crossentropy',
                  metrics=['accuracy'])

# Entraîner
history_fcn = model_fcn.fit(x_train_flat, y_train_cat,
                            epochs=10,
                            batch_size=32,
                            validation_split=0.1,
                            verbose=1)

# Évaluer
test_loss, test_accuracy = model_fcn.evaluate(x_test_flat, y_test_cat, verbose=0)
print(f"Test Accuracy: {test_accuracy:.4f}")
```

```python
# Preprocess the data for a Convolutional Neural Network:
    # Reshape the input data to the shape expected by a Conv2D layer (e.g., (60000, 28, 28, 1)).
    # Normalize the pixel values by dividing by 255.
    # One-hot encode the target labels using keras.utils.np_utils.to_categorical().

from tensorflow.keras.utils import to_categorical

# Rechargement des données MNIST pour s'assurer que tout est bien défini
from tensorflow.keras.datasets import mnist
(x_train_cnn, y_train_cnn), (x_test_cnn, y_test_cnn) = mnist.load_data()

# 1. Reshape : ajout d'une dimension "canal" (grayscale = 1 canal)
x_train_cnn = x_train_cnn.reshape(-1, 28, 28, 1)
x_test_cnn = x_test_cnn.reshape(-1, 28, 28, 1)

# 2. Normalisation des pixels (valeurs entre 0 et 1)
x_train_cnn = x_train_cnn / 255.0
x_test_cnn = x_test_cnn / 255.0

# 3. One-hot encoding des étiquettes
y_train_cnn_cat = to_categorical(y_train_cnn, num_classes=10)
y_test_cnn_cat = to_categorical(y_test_cnn, num_classes=10)

# Vérification des shapes
x_train_cnn.shape, y_train_cnn_cat.shape, x_test_cnn.shape, y_test_cnn_cat.shape
```

```python
# Build and train a Convolutional Neural Network:
    # Create a Sequential model.
    # Add Conv2D and MaxPool2D layers.
    # Add a Flatten layer.
    # Add Dense layers with appropriate activation functions.
    # Compile and train the model, similar to the Fully Connected Neural Network.

# Installation de TensorFlow pour permettre la construction et l'entraînement du modèle CNN
!pip install -q tensorflow

# Redémarrage automatique du noyau à faire manuellement si nécessaire dans certains environnements
# Rechargement après installation
import tensorflow as tf
from tensorflow.keras import layers, models


# Chargement des données MNIST
(x_train_cnn, y_train_cnn), (x_test_cnn, y_test_cnn) = mnist.load_data()

# Prétraitement
x_train_cnn = x_train_cnn.reshape(-1, 28, 28, 1).astype("float32") / 255.0
x_test_cnn = x_test_cnn.reshape(-1, 28, 28, 1).astype("float32") / 255.0
y_train_cnn_cat = to_categorical(y_train_cnn, num_classes=10)
y_test_cnn_cat = to_categorical(y_test_cnn, num_classes=10)

# Création du modèle CNN
model_cnn = models.Sequential([
    layers.Conv2D(32, kernel_size=(3,3), activation='relu', input_shape=(28,28,1)),
    layers.MaxPooling2D(pool_size=(2,2)),
    layers.Conv2D(64, kernel_size=(3,3), activation='relu'),
    layers.MaxPooling2D(pool_size=(2,2)),
    layers.Flatten(),
    layers.Dense(128, activation='relu'),
    layers.Dense(10, activation='softmax')  # 10 classes pour les chiffres de 0 à 9
])

# Compilation du modèle
model_cnn.compile(optimizer='adam',
                  loss='categorical_crossentropy',
                  metrics=['accuracy'])

# Entraînement du modèle
history_cnn = model_cnn.fit(
    x_train_cnn, y_train_cnn_cat,
    epochs=10,
    batch_size=64,
    validation_split=0.1,
    verbose=1
)

# Évaluation sur les données de test
test_loss_cnn, test_accuracy_cnn = model_cnn.evaluate(x_test_cnn, y_test_cnn_cat, verbose=0)
test_loss_cnn, test_accuracy_cnn
```
