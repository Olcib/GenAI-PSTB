    # Pourquoi ce choix ?

# Comparaison : Agent Hybride vs Agent Réactif (sur plusieurs aspects clés) :
    # Si l’agent était réactif (au lieu d’hybride)
        a) Gestion de la mémoire
        Agent réactif : aucune mémoire persistante — chaque interaction est traitée indépendamment.
        Cela signifie qu’il oublierait les symptômes déjà recueillis ou les décisions prises quelques secondes plus tôt.
        Conséquence : interactions redondantes, perte de contexte, confusion possible.
        b) Étapes de planification
        Agent réactif : agit uniquement en réponse à l'entrée immédiate, sans planification à long terme.
        Pas de raisonnement global ni de stratégie : il pourrait envoyer une ambulance dès un mot-clé sans tenir compte du contexte.
        Conséquence : sur-intervention ou sous-intervention fréquente.
        c) Stratégie d’appel aux outils
        Réactif : invoque les outils directement sur le moment, souvent sans coordonner plusieurs outils ensemble.
        Hybride : coordonne plusieurs outils dans une séquence logique (ex : parse → score → décision → action).

# 8. Reflect on Critical Questions
    # What fails if your agent does not maintain state?
    # Why are external tools (APIs, models) essential in an EMR dispatch scenario?
    # Write 2–3 sentences for each reflection question.

    # Que se passe-t-il si l’agent ne maintient pas d’état ?
      Si l’agent ne maintient pas d’état, il oublie les informations précédemment collectées (identité, symptômes, décisions déjà prises). Cela peut entraîner des questions redondantes, des erreurs de triage (ex : re-évaluation erronée), et une perte de temps précieuse en situation d’urgence. L’expérience utilisateur devient incohérente, voire dangereuse pour des patients en situation critique.

    # Pourquoi les outils externes (APIs, modèles) sont-ils essentiels dans un scénario de dispatch médical ?
      Les outils externes (comme les APIs de symptom checker, de triage médical ou de géolocalisation) apportent une expertise métier vitale que l’agent ne peut posséder en interne. Ils permettent de fiabiliser les décisions, de connecter en temps réel avec les secours, et de bénéficier de bases de données médicales à jour. Sans ces outils, l’agent fonctionnerait en vase clos, avec un niveau de compétence et de précision insuffisant pour un usage médical.
